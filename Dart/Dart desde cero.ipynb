# Aprendiendo Dart

Para el desarrollo de cualquier aplicacion en dart se requiere una funcion de nivel superior **main()**, donde se inicia la ejecucion. LAs funciones que no devuelven un valor explicito tiene  el **void** tipo de retorno. Por ejemplo para mostrar un texto en la consola se puede usar la funcion **print**

```dart
void main() {
  print('Hello, World!');
}
